**Zero - shot Audion Classification **

We are using CLAP(Contrastive Language Audio Pretraining)

```python
from transformers import pipeline
classifier=pipeline("zero-shot-audio-classification",model="laion/clap-htsat-fused")
audio_path="/content/recording.ogg"
labels=["Music","Bird singing","Human talking"]
result=classifier(audio_path, candidate_labels=labels, hypothesis_template="This is a sound of {}.")
print(f"Result: {result}")
```

```python
from moviepy.editor import AudioFileClip

input_mp4_file = "/content/kannada.mp4"
output_mp3_file = "/content/kannada.mp3"

# Load the audio from the MP4 file
audio_clip = AudioFileClip(input_mp4_file)

# Write the audio to an MP3 file
audio_clip.write_audiofile(output_mp3_file)

# Close the audio clip
audio_clip.close()

print(f"Successfully converted '{input_mp4_file}' to '{output_mp3_file}'")
```

```python
asr=pipeline("automatic-speech-recognition",model="openai/whisper-tiny", chunk_length_s=30)
transcription=asr("/content/kannada.mp3")
print(f"Transcription: {transcription}")
```

```python
from transformers import VitsModel, AutoTokenizer
import torch
import scipy

model_name="facebook/mms-tts-kan"
model=VitsModel.from_pretrained(model_name)
tokenizer=AutoTokenizer.from_pretrained(model_name)

text="ನಮಸ್ಕಾರ, ಈ ಸರ್ಕಾರಿ ಯೋಜನೆಯ ಮಾಹಿತಿ ಇಲ್ಲಿದೆ."
inputs=tokenizer(text,return_tensors="pt")
with torch.no_grad():
  output=model(**inputs).waveform
scipy.io.wavfile.write("output.wav",rate=16000,data=output.numpy().T)
```
