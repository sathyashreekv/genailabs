```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, Dot, Reshape, Dense

# Sample corpus
corpus = [
    "machine learning is fun",
    "deep learning is powerful",
    "natural language processing is interesting"
]

# Step 1: Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)
word_index = tokenizer.word_index
vocab_size = len(word_index) + 1

sequences = tokenizer.texts_to_sequences(corpus)

# Step 2: Manually create Skip-gram pairs
window_size = 2
pairs = []

for seq in sequences:
    for i, target_word in enumerate(seq):
        start = max(0, i - window_size)
        end = min(len(seq), i + window_size + 1)

        for j in range(start, end):
            if i != j:
                context_word = seq[j]
                pairs.append((target_word, context_word))

# Convert to arrays
word_target = np.array([pair[0] for pair in pairs])
word_context = np.array([pair[1] for pair in pairs])
labels = np.ones(len(word_target))  # positive samples only

# Step 3: Build Skip-Gram Model
embedding_dim = 50

input_target = Input((1,))
input_context = Input((1,))

embedding = Embedding(vocab_size, embedding_dim)

target_embed = Reshape((embedding_dim,))(embedding(input_target))
context_embed = Reshape((embedding_dim,))(embedding(input_context))

dot_product = Dot(axes=1)([target_embed, context_embed])
output = Dense(1, activation='sigmoid')(dot_product)

model = Model([input_target, input_context], output)
model.compile(loss='binary_crossentropy', optimizer='adam')

# Step 4: Train
model.fit([word_target, word_context], labels, epochs=100, verbose=1)

# Step 5: Extract word embeddings
word_embeddings = embedding.get_weights()[0]

print("Vocabulary:", word_index)
print("Embedding shape:", word_embeddings.shape)
```
